```python
import google.generativeai as genai
import gradio as gr

# Configure API key
genai.configure(api_key="AIzaSyDM1NuGAApez9lkYw9HNHXq1wfTVGyugvU")  # Replace with your actual Gemini API key

# Load the model
model = genai.GenerativeModel(model_name="gemini-3-flash-preview")

# System message (used only for context)
system_msg = """You are CineBot AI, an expert movie and anime review assistant.

Your responsibilities:
- Recommend movies and anime
- Explain reviews clearly
- Suggest based on genre and mood
- Keep answers engaging and concise
- Use emojis occasionally
- Sound friendly and intelligent

Format responses neatly."""

# Chatbot function
def CustomGemini(user_input):
    response = model.generate_content(system_msg + "\n" + user_input)
    return response.text  # Extract and return response

# Gradio Interface
demo = gr.Interface(fn=CustomGemini, 
                    inputs="text", 
                    outputs="text", 
                    title="Movie Review Assistent",
                    description="Ask for movie recommendations, reviews, anime suggestions, and ratings.",
                    examples=[
                            "Suggest action anime",
                            "Best Christopher Nolan movies",
                            "Recommend emotional movies",
                            "Top sci-fi thrillers",
                            "Best horror movies of all time"
                        ],
                    theme="soft")

# Launch the app
demo.launch(share=True)
```
